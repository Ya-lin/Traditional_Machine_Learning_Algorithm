### t-tests
t-tests (also known as Student's t-tests) are used to compare two mean (average) statistics and conclude whether they are different enough from each other.

```python
import numpy as np
from scipy import stats
from matplotlib import pyplot as plt
```

```python
sample_a = np.random.normal(size=20)
sample_b = np.random.normal(0.2, 1, size=20)

plt.hist(sample_a, alpha=0.2, label='sample-a')
plt.hist(sample_b, alpha=0.2, label='sample-b')
plt.legend()
plt.show()
```

In a t-test, the null hypothesis is that the two means are indeed equal. If the pvalue is larger than a fixed threshold (usually, $0.05$ is chosen), we cannot reject the hypothesis. If the pvalue is smaller than the threshold, we reject the null hypothesis, implying that the two means are different.

```python
pvalue = stats.ttest_ind(sample_a, sample_b).pvalue
print(pvalue)
```

With this result, we do not reject our hypothesis. We don't have enough evidence to say that the means of our two samples are different.

```python
sample_a = np.random.normal(size=1000)
sample_b = np.random.normal(0.2, 1, size=1000)

plt.hist(sample_a, alpha=0.2, label='sample-a')
plt.hist(sample_b, alpha=0.2, label='sample-b')
plt.legend()
plt.show()
```

```python
pvalue = stats.ttest_ind(sample_a, sample_b).pvalue
print(pvalue)
```

This pvalue is smaller than $0.05$, thus rejecting the null hypothesis and giving us enough evidence to say that the two samples have different means.
